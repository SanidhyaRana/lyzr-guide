## Step 6: Create Simple Gemini Model

Let's create our AI model wrapper that works with Lyzr Automata:

````python
# Simple Gemini Model for beginners
class SimpleGeminiModel(AIModel):
    def __init__(self, api_key):
        self.api_key = api_key
        genai.configure(api_key=api_key)
        self.model = genai.GenerativeModel("gemini-2.0-flash-exp")
        
    def generate_text(self, prompt, **kwargs):
        try:
            response = self.model.generate_content(prompt)
            return response.text
        except Exception as e:
            print(f"Error: {e}")
            return "Sorry, I couldn't generate a response."
    
    def generate_image(self, prompt, **kwargs):
        # This model doesn't generate images, return text description
        return f"Image description: {prompt}"
````

## Step 9: Create a Research Agent

An **Agent** is like giving the AI a specific role and expertise:

````python
# Create a research agent
research_agent = Agent(
    role="Senior Research Analyst",
    prompt_persona="You are an expert research analyst who conducts comprehensive analysis on any given topic. You provide detailed, accurate, and well-structured research findings."
)

print("Research agent created!")
print(f"Agent role: {research_agent.role}")
````

## Step 10: Create a Research Task

A **Task** tells the agent exactly what to do:

````python
# Create research task
def create_research_task(topic):
    return Task(
        name="Topic Analysis",
        agent=research_agent,
        model=gemini_model,
        instructions=f"Conduct a comprehensive analysis of '{topic}'. Include key points, current trends, and important facts. Make it informative and engaging."
    )
````

## Step 11: Create a Blog Writing Agent

Let's create another agent specialized in writing blogs:

````python
# Create blog writing agent
blog_writer_agent = Agent(
    role="Digital Content Creator",
    prompt_persona="You are an expert digital content creator who specializes in writing engaging, SEO-friendly blog posts. You create content that is informative, well-structured, and engaging for readers."
)

print("Blog writer agent created!")
print(f"Agent role: {blog_writer_agent.role}")
````

## Step 12: Create a Blog Writing Task

This task will write a complete blog post:

````python
# Create blog writing task
def create_blog_task(topic):
    return Task(
        name="Blog Content Creation",
        agent=blog_writer_agent,
        model=gemini_model,
        instructions=f"""
        Write a comprehensive blog post about '{topic}' with the following structure:
        
        1. Create an eye-catching title
        2. Write an engaging introduction
        3. Include 3-5 main sections with clear headings
        4. Add practical tips or insights
        5. Write a compelling conclusion
        6. Use markdown formatting for better readability
        
        Make it informative, engaging, and around 800-1000 words.
        """
    )

print(" Blog writing task function created!")
````

## Step 13: Create an MCQ Generator Agent

Let's create an agent that generates multiple choice questions:

````python

mcq_agent = Agent(
    role="Educational Content Expert",
    prompt_persona="You are an expert in creating educational content, specifically multiple choice questions (MCQs). You create clear, challenging, and educational questions with accurate answers."
)

print(" MCQ generator agent created!")
print(f"Agent role: {mcq_agent.role}")
````

## Step 14: Create MCQ Generation Task

This task will create multiple choice questions on any topic:

````python
# Create MCQ generation task
def create_mcq_task(topic):
    return Task(
        name="MCQ Generation",
        agent=mcq_agent,
        model=gemini_model,
        instructions=f"""
        Create 10 multiple choice questions about '{topic}' with the following format:
        
        For each question:
        1. Write a clear question
        2. Provide 4 answer options (A, B, C, D)
        3. Mark the correct answer
        4. Add a brief explanation for the correct answer
        
        Make sure questions are educational, varied in difficulty, and cover different aspects of the topic.
        """
    )
````

## Step 15: Simple Research Function

Let's create a simple function to run research analysis:

````python
def run_research_analysis(topic):

    
    # Create the research task
    research_task = create_research_task(topic)
    
    # Create and run the pipeline
    pipeline = LinearSyncPipeline(
        name="Research Analysis Pipeline",
        completion_message=f"Research analysis completed for: {topic}",
        tasks=[research_task]
    )
    
    result = pipeline.run()
    return result[0]['task_output'] if result else "No results generated"
````

## Step 16: Blog Generation Function

Let's create a function that generates complete blog posts:

````python
def generate_blog_post(topic):
    # Create research and blog tasks
    research_task = create_research_task(topic)
    blog_task = create_blog_task(topic)

    pipeline = LinearSyncPipeline(
        name="Blog Generation Pipeline",
        completion_message=f"Blog post completed for: {topic}",
        tasks=[research_task, blog_task]
    )
    
    result = pipeline.run()
    return result[1]['task_output'] if len(result) > 1 else "No blog generated"
````

## Step 17: MCQ Generation Function

Let's create a function for generating MCQ questions:

````python
def generate_mcq_questions(topic):

    mcq_task = create_mcq_task(topic)
  
    pipeline = LinearSyncPipeline(
        name="MCQ Generation Pipeline",
        completion_message=f"MCQ questions completed for: {topic}",
        tasks=[mcq_task]
    )
    
    result = pipeline.run()
    return result[0]['task_output'] if result else "No MCQs generated"
````

# 🤝 Multi-Agent Coordination Examples

Now let's explore how multiple agents can work together in Lyzr Automata! We'll use our existing agents to demonstrate different coordination patterns.

## Pattern 1: Sequential Agent Coordination

In this pattern, agents work one after another, where each agent builds upon the previous agent's work:

````python
def sequential_content_pipeline(topic):
    """
    Sequential coordination: Research → Blog → MCQ
    Each agent builds upon the previous agent's output
    """
    
    # Task 1: Research the topic
    research_task = Task(
        name="Research Phase",
        agent=research_agent,
        model=gemini_model,
        instructions=f"Research '{topic}' thoroughly. Provide key insights, trends, and important facts."
    )
    
    # Task 2: Write blog using research findings
    blog_task = Task(
        name="Blog Writing Phase", 
        agent=blog_writer_agent,
        model=gemini_model,
        instructions=f"Using the research findings, write a comprehensive blog post about '{topic}'. Include practical examples and make it engaging for readers."
    )
    
    # Task 3: Create MCQs based on blog content
    mcq_task = Task(
        name="MCQ Creation Phase",
        agent=mcq_agent, 
        model=gemini_model,
        instructions=f"Based on the blog content, create 5 multiple choice questions that test understanding of '{topic}'. Include explanations for correct answers."
    )
    
    # Sequential pipeline - agents work one after another
    pipeline = LinearSyncPipeline(
        name="Sequential Content Creation Pipeline",
        completion_message=f"Complete content package created for: {topic}",
        tasks=[research_task, blog_task, mcq_task]  # Sequential execution
    )
    
    results = pipeline.run()
    return results


topic = "Blockchain Technology"
sequential_results = sequential_content_pipeline(topic)
````

````python
# Display results from sequential coordination
print(" Sequential Pipeline Results:\n")


for i, result in enumerate(sequential_results, 1):
    task_name = result.get('task_name', f'Task {i}')
    output = result.get('task_output', 'No output')
    
    print(f"\n🎯 {task_name}:")
   
    print(output[:500] + "..." if len(output) > 500 else output)
````

## Pattern 2: Parallel Agent Coordination

In this pattern, multiple agents work simultaneously on different aspects of the same topic:

````python
def parallel_content_creation(topic):

    # Create multiple specialized agents for different perspectives
    technical_researcher = Agent(
        role="Technical Research Specialist",
        prompt_persona="You are a technical expert who focuses on the technical aspects, implementation details, and future prospects of technologies."
    )
    
    business_researcher = Agent(
        role="Business Analysis Expert", 
        prompt_persona="You are a business analyst who focuses on market impact, business applications, and economic implications of technologies."
    )
    
    educational_content_creator = Agent(
        role="Educational Content Specialist",
        prompt_persona="You are an educational expert who creates learning materials, explains complex concepts simply, and develops educational assessments."
    )
    
    # Parallel tasks - all run simultaneously 
    technical_task = Task(
        name="Technical Analysis",
        agent=technical_researcher,
        model=gemini_model,
        instructions=f"Analyze the technical aspects of '{topic}'. Cover implementation, architecture, challenges, and future developments."
    )
    
    business_task = Task(
        name="Business Impact Analysis",
        agent=business_researcher, 
        model=gemini_model,
        instructions=f"Analyze the business implications of '{topic}'. Cover market opportunities, ROI, adoption challenges, and business use cases."
    )
    
    educational_task = Task(
        name="Educational Content Creation",
        agent=educational_content_creator,
        model=gemini_model,
        instructions=f"Create educational content about '{topic}'. Include key learning objectives, beginner-friendly explanations, and practice questions."
    )
    
    # Note: In Lyzr's LinearSyncPipeline, tasks run sequentially
    # For true parallel execution, you'd run separate pipelines
    
    # Technical perspective pipeline
    tech_pipeline = LinearSyncPipeline(
        name="Technical Analysis Pipeline",
        completion_message="Technical analysis complete",
        tasks=[technical_task]
    )
    
    # Business perspective pipeline  
    business_pipeline = LinearSyncPipeline(
        name="Business Analysis Pipeline", 
        completion_message="Business analysis complete",
        tasks=[business_task]
    )
    
    # Educational perspective pipeline
    edu_pipeline = LinearSyncPipeline(
        name="Educational Content Pipeline",
        completion_message="Educational content complete", 
        tasks=[educational_task]
    )
    
    # Run all pipelines (simulating parallel execution)

    tech_results = tech_pipeline.run()
    business_results = business_pipeline.run() 
    edu_results = edu_pipeline.run()
    
    return {
        'technical': tech_results[0] if tech_results else None,
        'business': business_results[0] if business_results else None,
        'educational': edu_results[0] if edu_results else None
    }

# Test parallel coordination
topic = "Artificial Intelligence in Healthcare"
parallel_results = parallel_content_creation(topic)
````

````python
# Display parallel coordination results
print(" Parallel Agent Coordination Results:\n")


for perspective, result in parallel_results.items():
    if result:
        print(f"\n {perspective.upper()} PERSPECTIVE:")
        output = result.get('task_output', 'No output available')
        print(output)
    
````

## Pattern 3: Hierarchical Agent Coordination

In this pattern, we have a coordinator agent that manages other agents and synthesizes their outputs:

````python
def hierarchical_coordination(topic):

    
    # Create coordinator agent
    coordinator_agent = Agent(
        role="Project Coordinator & Content Synthesizer",
        prompt_persona="You are a project coordinator who manages multiple specialist teams, synthesizes their work, and creates unified deliverables that combine insights from all specialists."
    )
    
    # Step 1: Gather specialist insights (using existing agents)
    research_task = Task(
        name="Research Gathering",
        agent=research_agent,
        model=gemini_model, 
        instructions=f"Conduct comprehensive research on '{topic}'. Focus on factual information, current state, and key developments."
    )
    
    blog_task = Task(
        name="Content Strategy",
        agent=blog_writer_agent,
        model=gemini_model,
        instructions=f"Develop content strategy and key messaging for '{topic}'. Identify target audience and engagement approaches."
    )
    
    mcq_task = Task(
        name="Assessment Strategy", 
        agent=mcq_agent,
        model=gemini_model,
        instructions=f"Identify key learning objectives and assessment points for '{topic}'. Define what audiences should know."
    )
    
    # Step 2: Coordinator synthesizes all inputs
    synthesis_task = Task(
        name="Content Synthesis & Final Deliverable",
        agent=coordinator_agent,
        model=gemini_model,
        instructions=f"""
        As the project coordinator, synthesize the research findings, content strategy, and assessment insights about '{topic}' into a comprehensive project deliverable.
        
        Create a unified report that includes:
        1. Executive Summary 
        2. Key Findings & Insights
        3. Strategic Recommendations
        4. Content Framework
        5. Success Metrics
        
        Ensure all specialist inputs are integrated coherently into a single, actionable deliverable.
        """
    )
    
    # Hierarchical pipeline: specialists → coordinator
    pipeline = LinearSyncPipeline(
        name="Hierarchical Coordination Pipeline",
        completion_message=f"Hierarchical coordination completed for: {topic}",
        tasks=[research_task, blog_task, mcq_task, synthesis_task]
    )
    
    results = pipeline.run()
    return results

# Test hierarchical coordination
topic = "Sustainable Energy Solutions"
hierarchical_results = hierarchical_coordination(topic)
````

````python

print(" SPECIALIST CONTRIBUTIONS:")
for i, result in enumerate(hierarchical_results[:-1], 1):
    task_name = result.get('task_name', f'Specialist Task {i}')
    output = result.get('task_output', 'No output')
    print(f"\n {task_name}:")
    print("-" * 25)
    print(output)

# Show coordinator's synthesis
if len(hierarchical_results) > 3:
    coordinator_result = hierarchical_results[-1]
    print(f"\n\n COORDINATOR'S UNIFIED DELIVERABLE:")
    print("=" * 50)
    output = coordinator_result.get('task_output', 'No synthesis available')
    print(output)
````

## Pattern 4: Collaborative Review & Refinement

In this pattern, agents review and improve each other's work through multiple iterations:

````python
def collaborative_refinement(topic):

    # Create review agent
    review_agent = Agent(
        role="Quality Assurance & Content Reviewer",
        prompt_persona="You are a meticulous content reviewer who identifies areas for improvement, fact-checks information, and suggests enhancements to make content more accurate, engaging, and comprehensive."
    )
    
    # Round 1: Initial content creation
    initial_research = Task(
        name="Initial Research Draft",
        agent=research_agent,
        model=gemini_model,
        instructions=f"Create an initial research draft about '{topic}'. Focus on gathering comprehensive information."
    )
    
    # Round 2: Review initial research
    research_review = Task(
        name="Research Review & Feedback",
        agent=review_agent, 
        model=gemini_model,
        instructions=f"Review the research draft about '{topic}'. Identify gaps, inaccuracies, or areas that need more detail. Provide specific feedback for improvement."
    )
    
    # Round 3: Refined research based on feedback
    refined_research = Task(
        name="Refined Research",
        agent=research_agent,
        model=gemini_model,
        instructions=f"Based on the reviewer's feedback, create an improved and more comprehensive research report about '{topic}'. Address all identified gaps and suggestions."
    )
    
    # Round 4: Create blog based on refined research
    blog_creation = Task(
        name="Blog Creation", 
        agent=blog_writer_agent,
        model=gemini_model,
        instructions=f"Using the refined research, create an engaging blog post about '{topic}'. Make it informative and reader-friendly."
    )
    
    # Round 5: Review the blog
    blog_review = Task(
        name="Blog Review & Enhancement",
        agent=review_agent,
        model=gemini_model, 
        instructions=f"Review the blog post about '{topic}'. Suggest improvements for engagement, clarity, and completeness. Identify any missing elements."
    )
    
    # Round 6: Final refined blog
    final_blog = Task(
        name="Final Enhanced Blog",
        agent=blog_writer_agent,
        model=gemini_model,
        instructions=f"Create the final version of the blog post about '{topic}', incorporating all reviewer suggestions and enhancements."
    )
    
    # Collaborative pipeline with multiple review cycles
    pipeline = LinearSyncPipeline(
        name="Collaborative Refinement Pipeline",
        completion_message=f"Collaborative refinement completed for: {topic}",
        tasks=[initial_research, research_review, refined_research, blog_creation, blog_review, final_blog]
    )
    
    results = pipeline.run()
    return results

topic = "Quantum Computing Applications"
collaborative_results = collaborative_refinement(topic)
````

````python

# Group results by rounds
rounds = [
    ("Round 1 - Initial Research", collaborative_results[0] if len(collaborative_results) > 0 else None),
    ("Round 2 - Research Review", collaborative_results[1] if len(collaborative_results) > 1 else None), 
    ("Round 3 - Refined Research", collaborative_results[2] if len(collaborative_results) > 2 else None),
    ("Round 4 - Blog Creation", collaborative_results[3] if len(collaborative_results) > 3 else None),
    ("Round 5 - Blog Review", collaborative_results[4] if len(collaborative_results) > 4 else None),
    ("Round 6 - Final Blog", collaborative_results[5] if len(collaborative_results) > 5 else None)
]

for round_name, result in rounds:
    if result:
        print(f"\n {round_name}:")
        output = result.get('task_output', 'No output available')
 
        max_length = 200 if 'Review' in round_name else 400
        print(output)
        
````
